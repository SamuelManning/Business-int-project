```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('hour.csv')

df['dteday'] = pd.to_datetime(df['dteday'])

fig = px.scatter(
    df, 
    x='dteday', 
    y='cnt',          
    color='temp',         
    color_continuous_scale='Turbo', #Strange method for coloring
    title='Impact of Temperature on Bike Rentals Over Time',
    labels={'dteday': 'Date', 'cnt': 'Bike Rentals', 'temp': 'Temperature'}
)

fig.update_layout(
    xaxis=dict(
        rangeslider=dict(visible=True), 
        type='date'
    ),
    template='plotly_white'
)

fig.show()
```
